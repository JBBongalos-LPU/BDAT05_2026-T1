# §3 · The grain — Build the branch-month revenue grain  *(≈12 minutes)*

Using your repaired IDs: build a table where **one row = one branch in one month**, with total `Amount`. Exclude rows with no date, and say so. **Print the row count** of your grain table.


```python
# §3 — your grain code
# Step 1: Convert date to datetime and count rows with no date
orders["Date"] = pd.to_datetime(orders["Date"], format="%Y-%m-%d")

missing_date_count = orders["Date"].isna().sum()

print("Rows excluded due to missing date:", missing_date_count)
```

```python
# Step 2: Create one row per branch per month
orders_valid = orders.dropna(subset=["Date"]).copy()

orders_valid["Month"] = orders_valid["Date"].dt.to_period("M")

branch_month = (
    orders_valid
    .groupby(["BranchID", "Month"], as_index=False)["Amount"]
    .sum()
)

print("Rows in branch-month grain table:", len(branch_month))
```

# §5 · Judgment — Judgment — the ₱ outlier  *(≈14 minutes)*

Compute the IQR fences on `Amount`. Report how many rows sit beyond the upper fence — then look at your **single most extreme row** (its service, its weight, its amount) and RULE on it: **KEEP / FLAG / DROP**, with the reason a manager would accept. Numbers first, verdict second.


```python
# §5 — computation (if your task needs it)
# Step 1: Compute the IQR and upper fence for amount
q1 = orders["Amount"].quantile(0.25)
q3 = orders["Amount"].quantile(0.75)
iqr = q3 - q1
upper_fence = q3 + (1.5 * iqr)

upper_outliers = orders[orders["Amount"] > upper_fence]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Upper fence:", upper_fence)
print("Rows beyond upper fence:", len(upper_outliers))
```
