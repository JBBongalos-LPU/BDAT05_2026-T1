# Question 1 — The model you were told to build

A marketing agency in Santa Rosa wins a new client: a chain of milk-tea shops with about 900 registered customers. In the kickoff meeting, the owner says:

*"Predict which of my customers will stop buying next quarter. I read that this is easy now with AI."*

The analyst receives the CRM export the next morning. It contains only **current** customers — the shop's encoder deletes lapsed accounts twice a year *"to keep the file small."* Each remaining customer appears **exactly once**, with their sign-up date and their total lifetime spend.

> *A model learns to recognise leaving from the ones who left.*
> *A file with no leavers can only teach loyalty — and it will, confidently.*

### Answer all three parts

**(a)** Explain, in the language of this course, why this file **cannot train the model the owner asked for**. Your answer should use at least two of: *target variable, positive examples, unit of analysis, rows per entity*. Give one example from data you have personally worked with where the same wall appeared.

**(b)** The owner says: *"Just run the AI on it anyway."* Describe what you would propose to do **instead**, in order — and state **the specific test** that tells you the day real modeling becomes possible.

**(c)** Saying "not yet" to a paying client has a cost. So does shipping a model that cannot work. **Which costs more here, and why?** Then say what you would put **in writing** to the owner, and where that record lives.


### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_1 = """
(a)Given that the shop’s encoder has deleted historical customer’s data leaving only their sign-up date and total lifetime spending, this model alone could not accurately predict which customers will stop buying next quarter because the only data present is their lifetime spending and their sign-up date. Like in the case of Kape Tayo’s data, historical data shows when B01 (Calamba) branch opened, thus its sales data is proportional to how long it is open.

	This data alone would only give inaccurate predictions which could cost the business a lot of resources. The right data would be the customer’s actual daily, weekly, and monthly spending behavior in the span of those years starting when the customer initially registered to the company. Since the shop’s encoder deleted most of the to “keep a small file” the analyst could not produce an accurate model.

(b)AI could have a prediction for the said model, BUT that would still be inaccurate due to the lack of available data present. I would suggest that we could train the model to see which of these customers has the lowest spending behavior to the highest and profile them accordingly for the meantime.

	What we should do is to either regain access to the deleted customer data or manually check the company’s sales data and align them with the customerID. Once the historical data is present, we could proceed with predicting which customer is most likely to stop buying next quarter.

(c)Releasing a model that is not 100% accurate or functional would cost your company more resources than sticking to what I am suggesting in retrieving the lost data first before proceeding with your request.

  Inaccurate predictions are most likely to ruin a company’s integrity due to lack of relevant and working data.

"""

print(f"Question 1 — {_wc(ANSWER_1)} words")
```

   **Which error costs more here, and why?** Your answer should depend on this situation — night-before staffing, real crews, real trucks — not on a rule.


### YOUR ANSWER — Question 2
