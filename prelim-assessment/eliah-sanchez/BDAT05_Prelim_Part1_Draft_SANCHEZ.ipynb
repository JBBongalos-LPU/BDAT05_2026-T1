# Question 1 — The model you were told to build

A marketing agency in Santa Rosa wins a new client: a chain of milk-tea shops with about 900 registered customers. In the kickoff meeting, the owner says:

*"Predict which of my customers will stop buying next quarter. I read that this is easy now with AI."*

The analyst receives the CRM export the next morning. It contains only **current** customers — the shop's encoder deletes lapsed accounts twice a year *"to keep the file small."* Each remaining customer appears **exactly once**, with their sign-up date and their total lifetime spend.

> *A model learns to recognise leaving from the ones who left.*
> *A file with no leavers can only teach loyalty — and it will, confidently.*

### Answer all three parts

**(a)** Explain, in the language of this course, why this file **cannot train the model the owner asked for**. Your answer should use at least two of: *target variable, positive examples, unit of analysis, rows per entity*. Give one example from data you have personally worked with where the same wall appeared.

**(b)** The owner says: *"Just run the AI on it anyway."* Describe what you would propose to do **instead**, in order — and state **the specific test** that tells you the day real modeling becomes possible.

**(c)** Saying "not yet" to a paying client has a cost. So does shipping a model that cannot work. **Which costs more here, and why?** Then say what you would put **in writing** to the owner, and where that record lives.


### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_1 = """
(a) The file cannot train the model the owner asked for because the data that the file provides are merely to store data of customers. In addition, the data received only contains current customers, not including the old customers. If the shop's encoder deletes lapsed accounts twice a year to keep the file small, it is highly unlikely that what the client is asking for is possible. Even with the suggestion of the owner of using AI, the range of the data is still not enough for AI to be able to create a predicting model. Also, there is no positive example. There would be no useful information from the CRM export for it to be used as a predictive model.


(b) I would propose to the owner to provide me with a data that is has a record long enough for me or AI to be able to use predictive analysis on the customers' behavior. In order, I would check the data first if it is clean, if it's not, I would provide the owner a list of the missing data, etc. Once the data I need is provided, I would create an analysis based on their average spending behavior, etc.


(c) Shipping a model that does cannot work would be more costly, because first, in order for a project to be successful, you need proper planning: Initiation, Planning, Execution, Monitoring and Controlling, and the release of the project. On our project management subject, starting a project without proper planning and having a lot of adjustments while it is ongoing, it would cost more due to the sudden changes, expenses, etc. I would put in writing to the owner all the things that they should expect if I shipped a model that cannot work vs a model that was planned and executed properly.

"""

print(f"Question 1 — {_wc(ANSWER_1)} words")
```

# Question 2 — The predictor that was too good

A junior analyst at a logistics company builds a model to forecast **each truck's daily delivery volume**, so the depot can assign crews the night before. He is thrilled:

*"It's 98% accurate! The strongest predictor is `fuel_used_that_day` — makes sense, right? More fuel, more deliveries. The correlation is 0.97. We deploy Friday."*

You are the senior analyst. Deployment means the depot will staff **tonight** based on what the model says about **tomorrow**.

> *The exam was passed. The morning was failed.*
> *Between those two sentences is the entire discipline you are being examined on.*

### Answer all three parts

**(a)** Name what is wrong with this predictor, using the correct term and the **one question** that exposes it. Then give one example — from data you have personally worked with — of a predictor that failed the same question.

**(b)** Write what you would actually **say to the junior analyst** — three or four sentences, as you would say them. Include the **honest repair** that keeps his insight but fixes his timing, and **one thing you would refuse to let happen on Friday**.

**(c)** Two ways your intervention can go wrong. You could kill a legitimate predictor and lose real accuracy. Or you could let a cheating one ship and fail in production.

   **Which error costs more here, and why?** Your answer should depend on this situation — night-before staffing, real crews, real trucks — not on a rule.


### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_2 = """
(a) As stated on the example, the staff, crew are scheduled the night before. So they need to know the exact amount of fuel a truck uses for a day. The correlation is very high and there's nothing wrong with it. A data that I've recently worked is me being the Chief Financial Officer, when I try to predict the yearly revenue, I did not consider adding in the months when working on my analysis or my predictive model to predict the next year's revenue. Because if we put it in this example, knowing whhat days of the week that show the most fuel consumption in a day, it would be easier for scheduling the staff the night before.


(b) I would commend the junior analyst for identifying that the fuel consumption is tied to the workload. One honest repair would be for the junior analyst to make the predictor into a 7 day average fuel consumption so it would be easier to visualize and to tell what day of the week would it be best to deploy.


(c) Cheating one ship and fail in production would cost more because it is more of a risk or a gamble than a predictor. Losing a legitimate predictor would only eliminate the ability for you to predict things. But risking it by cheating one ship and fail production could cost you more. If staffs and crews are already deployed, that would cost you money and if there are no deliveries, that would be a waste of money and time for you and the crew.

"""

print(f"Question 2 — {_wc(ANSWER_2)} words")
```
