# §3 · The grain — Build the payment-month grain  *(≈12 minutes)*

Using your parsed dates: build a table where **one row = one payment method in one month**, with the **count of orders**. **Print the row count** of your grain table.


```python
# ============================================================
# §3 · PAYMENT-MONTH GRAIN
# ============================================================

# Show all column names
print("Columns in orders:")
for i, col in enumerate(orders.columns):
    print(i, repr(col))


# Find the payment-method column automatically
payment_candidates = [
    col for col in orders.columns
    if "payment" in col.lower()
]

print("\nPossible payment columns:")
print(payment_candidates)
```

```python
# ============================================================
# §3 · PAYMENT-MONTH GRAIN + EVIDENCE
# ============================================================

# Identify the payment column
payment_candidates = [
    col for col in orders.columns
    if "payment" in col.lower()
]

payment_col = payment_candidates[0]

# Create month from parsed Date
orders["Payment_Month"] = orders["Date"].dt.to_period("M")

# Build payment-month grain
payment_month = (
    orders
    .dropna(subset=["Date", payment_col])
    .groupby(["Payment_Month", payment_col])
    .size()
    .reset_index(name="Order_Count")
    .sort_values(["Payment_Month", payment_col])
    .reset_index(drop=True)
)

# Show grain
print("PAYMENT-MONTH GRAIN")
print("-" * 50)
display(payment_month.head(20))

# §3 EVIDENCE
S3_GRAIN_ROWS = len(payment_month)

print("\n§3 EVIDENCE")
print("-" * 50)
print("Grain rows:", S3_GRAIN_ROWS)
```

# §4 · Diagnostic — Diagnostic — the column that lies about its type  *(≈14 minutes)*

One numeric-in-meaning column is stored as text. **Find it, explain WHY it happened** (show the evidence), recover it, and report how many values caused it and the recovered maximum.


```python
# ============================================================
# §4 · DIAGNOSTIC — FIND NUMERIC-IN-MEANING COLUMN STORED AS TEXT
# ============================================================

print("=" * 60)
print("NUMERIC-IN-MEANING COLUMNS STORED AS TEXT")
print("=" * 60)

for col in orders.columns:
    # Only inspect text/object columns
    if orders[col].dtype == "object":

        # Try converting the column to numeric
        converted = pd.to_numeric(orders[col], errors="coerce")

        # Count non-missing original values
        original_nonmissing = orders[col].notna().sum()

        # Count values successfully converted to numbers
        numeric_values = converted.notna().sum()

        # Calculate percentage that can be interpreted as numeric
        if original_nonmissing > 0:
            numeric_pct = numeric_values / original_nonmissing * 100
        else:
            numeric_pct = 0

        print(
            f"{col}: {numeric_values}/{original_nonmissing} "
            f"values numeric ({numeric_pct:.1f}%)"
        )
```

```python
# ============================================================
# §4 EVIDENCE — automatically identify and report the culprit
# ============================================================

# Find object/text columns that are numeric in meaning
numeric_text_columns = []

for col in orders.columns:
    if orders[col].dtype == "object":

        converted = pd.to_numeric(orders[col], errors="coerce")

        original_nonmissing = orders[col].notna().sum()
        numeric_values = converted.notna().sum()

        if original_nonmissing > 0:
            numeric_pct = numeric_values / original_nonmissing * 100

            if numeric_pct >= 95:
                numeric_text_columns.append(col)


# Select the suspected culprit
culprit = numeric_text_columns[0]

# Get the numeric version
culprit_numeric = pd.to_numeric(
    orders[culprit],
    errors="coerce"
)

# Get a few example culprit values
culprit_values = orders[culprit].dropna().head(5).tolist()

# Find the recovered maximum
recovered_max = culprit_numeric.max()


# Build evidence statement
S4_EVIDENCE = (
    f"Column: {culprit} "
    f"· culprit values: {culprit_values} "
    f"· recovered max: {recovered_max}"
)

print(S4_EVIDENCE)
```

# §5 · Judgment — Judgment — the 2,000-kilo order  *(≈14 minutes)*

Compute the IQR fences on your recovered `KiloWeight`. Report how many rows sit beyond the upper fence — then look at your **single heaviest order** and RULE on it: **KEEP / FLAG / DROP**, with the reason a manager would accept. Numbers first, verdict second.


```python
# ============================================================
# §5 · JUDGEMENT — 2000-KILO ORDER / KiloWeight OUTLIER CHECK
# ============================================================

# 1. Recover KiloWeight as numeric
orders["KiloWeight_numeric"] = pd.to_numeric(
    orders["KiloWeight"],
    errors="coerce"
)

# 2. Calculate Q1 and Q3
Q1 = orders["KiloWeight_numeric"].quantile(0.25)
Q3 = orders["KiloWeight_numeric"].quantile(0.75)

# 3. Calculate IQR
IQR = Q3 - Q1

# 4. Calculate IQR fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

# 5. Find rows beyond the upper fence
upper_outliers = orders[
    orders["KiloWeight_numeric"] > upper_fence
]

# 6. Find the single heaviest order
heaviest_index = orders["KiloWeight_numeric"].idxmax()
heaviest_order = orders.loc[[heaviest_index]]


# ============================================================
# RESULTS
# ============================================================

print("=" * 60)
print("KILOWEIGHT IQR DIAGNOSTIC")
print("=" * 60)

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower fence: {lower_fence}")
print(f"Upper fence: {upper_fence}")

print()
print("Rows beyond upper fence:", len(upper_outliers))

print()
print("=" * 60)
print("SINGLE HEAVIEST ORDER")
print("=" * 60)

display(heaviest_order)
```
