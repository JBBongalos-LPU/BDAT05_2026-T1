# §3 · The grain — Build the service-month revenue grain  *(≈12 minutes)*

On your deduplicated data: build a table where **one row = one service type in one month**, with total `Amount`. Exclude rows with no date, and say so. **Print the row count** of your grain table.


```python
orders["Date"] = pd.to_datetime(orders["Date"], errors="coerce")

orders = orders.dropna(subset=["Date"])

orders["Month"] = orders["Date"].dt.to_period("M")

grain = orders.groupby(["ServiceType", "Month"], as_index=False)["Amount"].sum()

print("Grain row count:", len(grain))
```

```python
# Create service-month grain
orders["Date"] = pd.to_datetime(orders["Date"], errors="coerce")
orders = orders.dropna(subset=["Date"])

orders["Month"] = orders["Date"].dt.to_period("M")

grain = orders.groupby(
    ["ServiceType", "Month"],
    as_index=False
)["Amount"].sum()

# Display grain table
display(grain)

# Show row count
print("Grain row count:", len(grain))
```

# §4 · Diagnostic — Diagnostic — the orphan branches  *(≈14 minutes)*

Compare the `BranchID` values in orders against the official `branches` file. **List every orphan value and count the stranded rows.** Then read your own list carefully: are all these orphans the *same kind* of problem? Say what you see.


```python
# Get official BranchIDs
official_branches = set(branches["BranchID"].dropna().astype(str).str.strip().str.upper())

# Standardize BranchIDs in orders for comparison
orders["BranchID_check"] = orders["BranchID"].astype(str).str.strip().str.upper()

# Find orphan BranchID values
orphans = orders[~orders["BranchID_check"].isin(official_branches)]

# List every orphan value and count stranded rows
print("=== ORPHAN BRANCH IDs ===")
print(orphans["BranchID_check"].value_counts())

print("\nTotal stranded rows:", len(orphans))
```

```python
## Find orphan BranchID values
official_branches = set(branches["BranchID"].astype(str).str.strip().str.upper())
order_branches = orders["BranchID"].astype(str).str.strip().str.upper()

orphan_values = order_branches[~order_branches.isin(official_branches)].unique()
stranded_rows = (~order_branches.isin(official_branches)).sum()

S4_EVIDENCE = f"Distinct orphan values: {len(orphan_values)} · stranded rows: {stranded_rows}"

print(S4_EVIDENCE)
```

# §5 · Judgment — Judgment — the honest average  *(≈14 minutes)*

Compute the **mean and the median** of `Amount`, and the skewness. The owner asks: *"what does a typical order earn us?"* Which number do you give her, and what would the other one get wrong? Numbers first, verdict second.


```python
# Compute mean, median, and skewness
mean_amount = orders["Amount"].mean()
median_amount = orders["Amount"].median()
skewness = orders["Amount"].skew()

print(f"Mean: ₱{mean_amount:,.2f}")
print(f"Median: ₱{median_amount:,.2f}")
print(f"Skewness: {skewness:.3f}")
```
