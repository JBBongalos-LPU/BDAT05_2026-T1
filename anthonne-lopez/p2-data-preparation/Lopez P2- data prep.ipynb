This slice uses sales transaction data containing BranchID, Date, and Amount to calculate total revenue by branch and month. It does not identify individual customers because the results are aggregated at the branch-month level and do not include customer names or CustomerID. Aggregating the data helps protect individual customer information because individual transactions are not shown in the final output. The analysis is intended only for business analysis and reporting purposes.

```python
# P4 — Create cleaned branch-month revenue dataset

import pandas as pd

# Load the original files
sales = pd.read_csv('/content/drive/MyDrive/sales_transactions.csv')
branches = pd.read_csv('/content/drive/MyDrive/branches.csv')

# -----------------------------
# 1. Clean BranchID formatting
# -----------------------------
sales["BranchID"] = (
    sales["BranchID"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.replace("-", "", regex=False)
)

branches["BranchID"] = (
    branches["BranchID"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.replace("-", "", regex=False)
)

# Keep only branch records with a BranchID
branches_clean = branches.dropna(subset=["BranchID"]).copy()

# Remove duplicate branch IDs if any
branches_clean = branches_clean.drop_duplicates(subset=["BranchID"])

# -----------------------------
# 2. Convert Date
# -----------------------------
sales["Date"] = pd.to_datetime(sales["Date"], errors="coerce")

# -----------------------------
# 3. Create Year and Month
# -----------------------------
sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month

# -----------------------------
# 4. Merge Sales + Branches
# -----------------------------
merged_p4 = pd.merge(
    sales,
    branches_clean[["BranchID", "BranchName", "City", "Province"]],
    on="BranchID",
    how="inner"
)

# -----------------------------
# 5. Create branch-month revenue
# -----------------------------
branch_month_revenue = (
    merged_p4
    .dropna(subset=["Year", "Month"])
    .groupby(
        ["BranchID", "BranchName", "City", "Province", "Year", "Month"],
        as_index=False
    )["Amount"]
    .sum()
)

# -----------------------------
# 6. Display result
# -----------------------------
print("Rows:", len(branch_month_revenue))
print("Branches:", branch_month_revenue["BranchID"].nunique())

branch_month_revenue.head(10)
```

```python
# Save the cleaned P4 dataset to Google Drive

output_path = '/content/drive/MyDrive/P4_branch_month_revenue_cleaned.csv'

branch_month_revenue.to_csv(output_path, index=False)

print("Cleaned dataset saved to:")
print(output_path)
```
