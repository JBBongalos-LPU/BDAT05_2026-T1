```python
def validate(df, name):

    print(f"=== {name} ===")
    print("rows:", len(df))
    print("dtypes:\n", df.dtypes)
    print("missing:\n", df.isna().sum())
    print("duplicate ids:", df["SaleID"].duplicated().sum())
    print("negative amounts:", (df["Amount"] < 0).sum())

    orphans = set(df["BranchID"]) - set(branches["BranchID"])
    print("orphan BranchIDs:", orphans)
```

```python
validate(sales, "Sales")
```

**Output**- The Sales table has 23 duplicate SaleIDs, which is a uniqueness problem. It also has 4,775 missing CustomerIDs, meaning many sales do not have a customer recorded. There are also 6 orphan BranchIDs caused by inconsistent formatting, such as lowercase letters, extra spaces, and hyphens. No negative amounts were found.

```python
sales["Date"] = pd.to_datetime(sales["Date"], errors="coerce")
```

```python
sales["Month"] = sales["Date"].dt.to_period("M")
```

```python
sales["BranchID"] = (
    sales["BranchID"]
    .astype(str)
    .str.strip()
    .str.upper()
    .str.replace("-", "", regex=False)
)

branches["BranchID"] = (
    branches["BranchID"]
    .astype(str)
    .str.strip()
    .str.upper()
    .str.replace("-", "", regex=False)
)
```

```python
merged = pd.merge(
    sales,
    branches,
    on="BranchID",
    how="inner"
)
```

```python
branch_month_revenue = (
    merged
    .groupby(["BranchID", "Month"])["Amount"]
    .sum()
    .reset_index()
)
```
