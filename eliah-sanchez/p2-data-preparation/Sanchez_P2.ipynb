```python
def validate(df, name):
    print(f"=== {name} ===")
    print("rows:", len(df))
    print("dtypes:\n", df.dtypes)
    print("missing:\n", df.isna().sum())
    print("duplicate ids:", df["SaleID"].duplicated().sum())
    print("negative amounts:", (df["Amount"] < 0).sum())
```

```python
validate(sales, "Sales")
```

```python
sales["Date"] = pd.to_datetime(sales["Date"])

sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month

product_year = sales.groupby(["ProductID", "Year"]).size()
product_month = sales.groupby(["ProductID", "Year", "Month"]).size()

print("Product-year rows:", len(product_year))
print("Product-month rows:", len(product_month))
```

```python
def validate_p5(df, name):
    print(f"=== {name} ===")
    print("rows:", len(df))
    print("dtypes:\n", df.dtypes)
    print("missing:\n", df.isna().sum())
    print("duplicate product-month rows:", df.duplicated(["ProductID", "Year", "Month"]).sum())
    print("negative amounts:", (df["Amount"] < 0).sum())
```

```python
product_month = (
    sales.groupby(["ProductID", "Year", "Month"])["Amount"]
    .sum()
    .reset_index()
)
```

```python
product_month.info()
```

```python
print("=== Product-Month Revenue ===")
print("rows:", len(product_month))
print("dtypes:\n", product_month.dtypes)
print("missing:\n", product_month.isna().sum())
print("duplicate product-month rows:", product_month.duplicated(["ProductID", "Year", "Month"]).sum())
print("negative amounts:", (product_month["Amount"] < 0).sum())
```

This table uses ProductID, Year, Month, and Amount from the sales data. It does not identify any person because it does not show names or CustomerID. The data is grouped by product and month, so individual customers cannot be identified. This makes the data more private.
