## Question 2 — The protocol under pressure  *(10 points)*

**TEAM 1.** A teammate, chasing a deadline, proposes: *"Tune the LightGBM on the test rounds — it's only eight
rounds, and the AUC will look better for the paper."* Refuse, in writing: **(a)** explain the 23 / 8 / 8 temporal
split and what each block is FOR, including how many times the test block may be touched and for what;
**(b)** explain why accuracy is not computed on your 4-percent outcome and what is read instead, in what order;
**(c)** explain what isotonic calibration adds, which block it is fitted on, where the reliability diagram is
drawn, and why a well-ranked but badly calibrated score cannot set a threshold.

**TEAM 2.** A reviewer writes: *"Your damped rule scored a median MASE of 0.729 at one year. Why should that
impress me?"* Answer, in writing: **(a)** define MASE and say exactly what 0.729 means in plain language, and why
the baseline lives inside the metric; **(b)** explain why the baselines were fitted BEFORE any model, and what the
paper would honestly report if nothing beat them; **(c)** explain rolling-origin evaluation and why a random
80/20 split on your panel would have produced a beautiful and worthless score.

```python
Q2_ANSWER = """
(a) We should not tune the LightGBM model using the test rounds because the test data must remain separate from model development. The 23 rounds are divided into three blocks: the first 8 rounds are used for validation, the middle 7 rounds are used for training, and the final 8 rounds are used for testing the final model. The validation block is used to select the best model and settings, while the test block should only be used once, after model selection and tuning are complete. Its purpose is to measure how well the final model performs on new, unseen data. If we repeatedly check the test results and make changes to the model based on them, the test data is no longer an unbiased measure of final model performance.

(b) We should not use accuracy as the main evaluation metric because the outcome occurs in only about 4% of cases. This means the outcome is highly imbalanced, with only a small proportion of cases being positive. A model could achieve high accuracy simply by predicting that the outcome will not occur for most cases, even if it performs poorly at identifying the 4% of positive cases. Instead, we should first evaluate AUC to determine how well the model separates the two outcome groups across different thresholds. We then evaluate calibration to determine whether the predicted probabilities are close to the observed outcomes. Therefore, the results should be interpreted in this order: AUC first, calibration second, and threshold selection last.

(c) Isotonic calibration helps make the model's predicted probabilities more accurate. A model can rank people correctly but still give probabilities that are too high or too low. We fit the isotonic calibration using the 8 validation rounds. Then, we use the 8 test rounds to make the reliability diagram. The diagram shows if the predicted probabilities are close to what actually happened. A model can have a good AUC but still have bad calibration. This means the model can rank people correctly, but its probability numbers may not be correct. Because of this, we should not use the score to set a threshold until we know that the probabilities are properly calibrated.

"""
print("Q2 words:", len(Q2_ANSWER.split()))
```

## Question 3 — The leak that whispers and the leak that shouts  *(10 points · both teams)*

In Hunt the Leak, `AvgTicket` sat at r = .05 — and neither of you caught it, because neither of you ran the
correlation matrix. In the Rev. 1 employment study, `hours_worked` DOMINATED the feature-importance plot — and it

**(a)** Explain how a near-zero correlation and a towering feature importance can BOTH be leaks — and state the
one test that catches both, in your own words. **(b)** Design the audit you will run on every candidate feature
list from P4 onward: the two steps, in order, and what each step can and cannot detect. **(c)** "Nothing changes
silently" was a cleaning rule in Session 3. Extend it to features: where does a feature decision — kept or
excluded — get recorded in your own track project, and what must the record contain?

```python
Q3_ANSWER = """
(a) A feature can still be a data leak even if it has very low correlation with the outcome. For example, AvgTicket had a correlation of only r = .05, which is very low, but it could still be a leak if it uses information that would not have been available at the time the prediction was made. Similarly, hours_worked was identified as a leak even though it had very high feature importance. This shows that correlation and feature importance cannot be used to determine whether a feature is a leak. The key test is to check when the information became available. If a feature uses information that comes from the future or would only be known after the prediction point, it is considered a data leak.

(b) For every feature list, we should perform two checks in order. First, we should check when the information was available and ask whether we would already have had that information at the time the prediction was being made. If the information came from the future, we should remove the feature because it introduces data leakage. However, this check only tells us whether a feature is valid to use; it does not tell us whether the feature is actually useful for prediction.

(c) The rule “nothing changes silently” should also be used for our features. Every feature that we keep or remove should be written in our project record. We should write the feature name, if we kept or removed it, and the reason why. We should also write the date and version. For example, if we remove hours_worked because it uses future information, we should write that in the record. This makes it clear what we changed and why we changed it.

"""
print("Q3 words:", len(Q3_ANSWER.split()))
```
